```python
# Write a program to convert RGB (Color) Image to grayscale image.
# Case 1: By taking mean average of three planes.
# Case 2: Input the weightage of three planes, i.e., R, G, B from the user. The weightage
# is a value between 0 and 1, and sum of all three weightages is equal to 1. (Example:
# Weightage of R, G, B is 0.7, 0.2, 0.1, respectively.)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read color image
img = cv2.imread("landscape.jpg")

if img is None:
    print("Image not found!")
else:
    # Convert BGR to RGB
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Separate RGB planes
    R = img[:, :, 0]
    G = img[:, :, 1]
    B = img[:, :, 2]

    # Calculate mean average
    gray = (R + G + B) / 3

    # Convert to uint8
    gray = gray.astype(np.uint8)

    # Display images
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img)
    plt.title("Original RGB Image")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(gray, cmap='gray')
    plt.title("Grayscale Image")
    plt.axis('off')

    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read color image
img = cv2.imread("landscape.jpg")

if img is None:
    print("Image not found!")
else:
    # Convert BGR to RGB
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Separate RGB planes
    R = img[:, :, 0]
    G = img[:, :, 1]
    B = img[:, :, 2]

    # Take weightage from user
    wr = float(input("Enter Red weightage: "))
    wg = float(input("Enter Green weightage: "))
    wb = float(input("Enter Blue weightage: "))

    # Check weightage conditions
    if (0 <= wr <= 1 and 0 <= wg <= 1 and 0 <= wb <= 1
            and np.isclose(wr + wg + wb, 1.0)):

        # Calculate grayscale image
        gray = (R * wr) + (G * wg) + (B * wb)

        # Convert to uint8
        gray = gray.astype(np.uint8)

        # Display images
        plt.figure(figsize=(10, 5))

        plt.subplot(1, 2, 1)
        plt.imshow(img)
        plt.title("Original RGB Image")
        plt.axis('off')

        plt.subplot(1, 2, 2)
        plt.imshow(gray, cmap='gray')
        plt.title("Grayscale Image")
        plt.axis('off')

        plt.show()

    else:
        print("Invalid weightage! Values must be between 0 and 1")
        print("Sum of all weightages must be equal to 1")
```
