```python
# Assignment 3
# 1. WAP to draw a border around the input image (or in other words pad the input image).
#  Input the width of the border (in terms of pixel count) and the desired color of the
# border from the user.
#  Try this for both binary and grayscale image.
# grayscale
import numpy as np
import matplotlib.pyplot as plt

# Read image in grayscale
image = cv2.imread("landscape.jpg", cv2.IMREAD_GRAYSCALE)

# Check if image is loaded
if image is None:
    print("Image not found. Check the image path.")
else:
    # Input border width
    width = int(input("Enter border width in pixels: "))

    # Input border color
    color = int(input("Enter border color (0-255): "))

    # Add border
    bordered_image = cv2.copyMakeBorder(
        image,
        width,
        width,
        width,
        width,
        cv2.BORDER_CONSTANT,
        value=color
    )

    # Display
    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(image, cmap="gray")
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(bordered_image, cmap="gray")
    plt.title("Image with Border")
    plt.axis("off")

    plt.show()
```

```python
# binary
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
img = cv2.imread("landscape.jpg", 0)

if img is None:
    print("Image not found!")

else:
    # Convert grayscale to binary
    _, binary = cv2.threshold(
        img, 127, 255, cv2.THRESH_BINARY
    )

    # Input border width and color
    width = int(input("Enter border width: "))
    color = int(input("Enter border color (0 or 255): "))

    if width < 0 or color not in [0, 255]:
        print("Invalid input!")
    else:
        # Add border
        border = cv2.copyMakeBorder(
            binary, width, width, width, width,
            cv2.BORDER_CONSTANT, value=color
        )

        # Display images
        plt.figure(figsize=(10, 5))

        plt.subplot(1, 2, 1)
        plt.imshow(binary, cmap='gray')
        plt.title("Original Binary Image")
        plt.axis('off')

        plt.subplot(1, 2, 2)
        plt.imshow(border, cmap='gray')
        plt.title("Binary Image with Border")
        plt.axis('off')

        plt.show()
```

```python
# WAP for Image Compliment. Try this for both binary and grayscale image. 
# grayscale
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
img = cv2.imread("landscape.jpg", 0)

if img is None:
    print("Image not found!")

else:
    # Calculate image complement
    complement = 255 - img

    # Display images
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img, cmap='gray')
    plt.title("Original Grayscale Image")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(complement, cmap='gray')
    plt.title("Complemented Image")
    plt.axis('off')

    plt.show()
```

```python
# binary image
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
img = cv2.imread("landscape.jpg", 0)

if img is None:
    print("Image not found!")

else:
    # Convert grayscale image to binary
    _, binary = cv2.threshold(
        img, 127, 255, cv2.THRESH_BINARY
    )

    # Calculate binary image complement
    complement = 255 - binary

    # Display images
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(binary, cmap='gray')
    plt.title("Original Binary Image")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(complement, cmap='gray')
    plt.title("Complemented Binary Image")
    plt.axis('off')

    plt.show()
```
